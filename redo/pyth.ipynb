```python
# Create a collection to store the current _id value
counter_collection = db['counters']

# Initialize the counter if it doesn't exist
if counter_collection.count_documents({'_id': 'productid'}) == 0:
    counter_collection.insert_one({'_id': 'productid', 'sequence_value': 0})

# Function to get the next _id value
def get_next_sequence_value(sequence_name):
    result = counter_collection.find_one_and_update(
        {'_id': sequence_name},
        {'$inc': {'sequence_value': 1}},
        return_document=True
    )
    return result['sequence_value']

# Insert a new document with auto-incremented _id
new_document = {
    '_id': get_next_sequence_value('productid'),
    'name': 'New Product',
    'price': 100
}
collection.insert_one(new_document)
print(f"Inserted document with _id: {new_document['_id']}")
```

```python
# Fetch all documents in the collection
documents = list(collection.find())

# Update each document with a new _id
for idx, doc in enumerate(documents, start=1):
    collection.update_one({'_id': doc['_id']}, {'$set': {'_id': idx}})

print("Updated _id for all documents.")
```
